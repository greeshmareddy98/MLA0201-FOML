```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
data = {
    'Income': [25000, 50000, 30000, 80000, 60000, 20000],
    'Loan': [20000, 10000, 25000, 15000, 12000, 30000],
    'Payment': [1, 0, 1, 0, 0, 1],
    'Credit': [0, 1, 0, 1, 1, 0]
}
df = pd.DataFrame(data)
X = df[['Income', 'Loan', 'Payment']]
y = df['Credit']
model = DecisionTreeClassifier()
model.fit(X, y)
result = model.predict([[55000, 12000, 0]])
if result[0] == 1:
    print("Credit Score: Good")
else:
    print("Credit Score: Poor")
```
