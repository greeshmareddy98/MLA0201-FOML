```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
X = np.array([
    [1000, 2],
    [1500, 3],
    [800, 2],
    [1200, 3],
    [2000, 4]
])
y = np.array([50, 75, 40, 60, 90])
model = LinearRegression()
model.fit(X, y)
prediction = model.predict([[1600, 3]])
print("Predicted House Price:", round(prediction[0], 2), "Lakhs")
plt.scatter(X[:, 0], y, label="Actual Price")
plt.scatter([1600], [prediction[0]], color="red", label="Predicted Price")
plt.xlabel("Area (sq.ft)")
plt.ylabel("Price (Lakhs)")
plt.title("House Price Prediction")
plt.legend()
plt.show()
```
