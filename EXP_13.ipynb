```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
X = np.array([
    [1, 10000],
    [2, 20000],
    [3, 30000],
    [4, 40000],
    [5, 50000]
])
y = np.array([900000, 800000, 700000, 600000, 500000])
model = LinearRegression()
model.fit(X, y)
prediction = model.predict([[2.5, 25000]])
print("Predicted Car Price: Rs.", prediction[0])
plt.scatter(X[:, 0], y, label="Actual Price")
plt.plot(X[:, 0], model.predict(X), label="Predicted Price")
plt.xlabel("Car Age")
plt.ylabel("Car Price (Rs.)")
plt.title("Car Price Prediction")
plt.legend()
plt.show()
```
