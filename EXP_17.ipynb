```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.ensemble import RandomForestRegressor
data = {
    'RAM': [2, 3, 4, 6, 8, 12],
    'Battery': [3000, 3500, 4000, 4500, 5000, 5500],
    'Price': [8000, 10000, 13000, 18000, 25000, 35000]
}
df = pd.DataFrame(data)
X = df[['RAM', 'Battery']]
y = df['Price']
model = RandomForestRegressor(n_estimators=100, random_state=42)
model.fit(X, y)
prediction = model.predict([[8, 5000]])
print("Predicted Mobile Price: Rs.", round(prediction[0], 2))
plt.scatter(df['RAM'], df['Price'])
plt.xlabel("RAM (GB)")
plt.ylabel("Mobile Price (Rs.)")
plt.title("Mobile Price Prediction")
plt.show()
```
